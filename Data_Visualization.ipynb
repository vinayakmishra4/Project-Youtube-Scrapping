# Data Visuilzation on the Youtube vedios dataset


```python
# STEP:-1 loading the libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# STEP:-2 loading the dataset
df=pd.read_excel('youtube_videos.xlsx')
df.head(5)
```

```python
# STEP:-3 data preprocessing
df.info()
df.describe()
df.isnull().sum()
df.dropna(inplace=True)
```

```python
# STEP 4: change the data type of the 'published_date' column to datetime
df["upload_date"] = pd.to_datetime(df['upload_date'], errors='coerce')
```

```python
# STEP 5: Data Visualization
# 1. Distribution of video views
plt.figure(figsize=(10, 6))
plt.hist(df['views'], bins=30, color='blue', edgecolor='black')
plt.title('Distribution of Video Views')
plt.xlabel('Number of Views')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# 2. Relationship between views and likes
plt.figure(figsize=(10, 6))
plt.scatter(df['views'], df['likes'], alpha=0.5)
plt.title('Relationship between Views and Likes')
plt.xlabel('Number of Views')
plt.ylabel('Number of Likes')
plt.grid()
plt.show()
```

```python
# 3. Top 10 most viewed videos
top_videos = df.sort_values(by='views', ascending=False).head(10)
plt.figure(figsize=(12, 6))
plt.barh(top_videos['title'], top_videos['views'], color='green')
plt.title('Top 10 Most Viewed Videos')
plt.xlabel('Number of Views')
plt.ylabel('Video Title')
plt.gca().invert_yaxis()  # Invert y-axis to show the highest views at the top
plt.show()
```

```python
# 4 . Upload date distribution
plt.figure(figsize=(10, 6))
plt.hist(df['upload_date'], bins=30, color='orange', edgecolor='black')
plt.title('Distribution of Upload Dates')
plt.xlabel('Upload Date')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# 5. Relationship between upload date and views
plt.figure(figsize=(10, 6))
plt.scatter(df['upload_date'], df['views'], alpha=0.5)
plt.title('Relationship between Upload Date and Views')
plt.xlabel('Upload Date')
plt.ylabel('Number of Views')
plt.grid()
plt.show()
```
